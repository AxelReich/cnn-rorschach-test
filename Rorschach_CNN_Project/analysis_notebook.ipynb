```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

# --- Configuration ---
# Where your results are saved
RESULTS_DIR = 'experiment_results'
DATA_DIR = 'data/rorschach_blots'

# The list of animals (our Decoder Key)
CLASS_NAMES = {
    0: 'Butterfly', 1: 'Cat', 2: 'Chicken', 3: 'Cow', 4: 'Dog',
    5: 'Elephant', 6: 'Horse', 7: 'Sheep', 8: 'Spider', 9: 'Squirrel'
}

def show_comparison(card_filename):
    """
    Displays the Original Inkblot, Model A's Vision, and Model B's Vision side-by-side.
    """
    # Construct file paths
    # Note: We look for the exact filenames your script generated
    original_path = os.path.join(DATA_DIR, card_filename)
    model_a_path = os.path.join(RESULTS_DIR, f"{card_filename}_ModelA.jpg")
    model_b_path = os.path.join(RESULTS_DIR, f"{card_filename}_ModelB.jpg")
    
    # Check if files exist before trying to load
    if not os.path.exists(model_a_path) or not os.path.exists(model_b_path):
        print(f"Skipping {card_filename}: Result images not found.")
        return

    # Create a figure with 3 subplots
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    
    # 1. Original Image
    img = mpimg.imread(original_path)
    axes[0].imshow(img)
    axes[0].set_title(f"Original: {card_filename}", fontsize=14, fontweight='bold')
    axes[0].axis('off')

    # 2. Model A (Naive)
    img_a = mpimg.imread(model_a_path)
    axes[1].imshow(img_a)
    axes[1].set_title("Model A (Naive)\nSees: [Check Image Label]", fontsize=12, color='red')
    axes[1].axis('off')

    # 3. Model B (Expert)
    img_b = mpimg.imread(model_b_path)
    axes[2].imshow(img_b)
    axes[2].set_title("Model B (Expert)\nSees: [Check Image Label]", fontsize=12, color='green')
    axes[2].axis('off')

    plt.show()

# --- Run the Analysis ---
print("=== Rorschach AI Experiment Results ===")
print("Comparing the 'Naive' Brain vs. the 'Expert' Brain\n")

# Get list of original cards
cards = sorted([f for f in os.listdir(DATA_DIR) if f.endswith(('.jpg', '.jpeg', '.png', '.webp'))])

for card in cards:
    show_comparison(card)
```
